### Question 1 [Easy - 2D Line Plot]
Plot a simple line chart showing the total Runs scored by the top 10 players in IPL 2024 (ordered by POS). Add an appropriate title and axis labels.

# Write your code below

### Question 2 [Easy - 2D Line Plot]
Plot the trend of the total number of content titles (Movies and TV Shows combined) released per year on Netflix between 2010 and 2021.

```python
# Write your code below
```

````python
# Filter years 2010–2021 and count titles per year
netflix_total = netflix[
    (netflix['release_year'] >= 2010) &
    (netflix['release_year'] <= 2021)
]['release_year'].value_counts().sort_index()

# Line Plot
plt.figure(figsize=(10,5))
plt.plot(netflix_total.index, netflix_total.values, marker='o')

plt.title('Total Netflix Content Released per Year (2010-2021)')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.xticks(netflix_total.index, rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
````

### Question 3 [Easy - 2D Line Plot]
Create a line plot showing the cumulative sum of Runs scored by the top 15 players in the IPL 2024 dataset.

```python
# Write your code below
```

````python
top15 = ipl.sort_values('Runs', ascending=False).head(15)
top15['cum_runs'] = top15['Runs'].cumsum()
top15[['Player', 'Runs', 'cum_runs']]

plt.plot(top15['Player'],top15['cum_runs'],marker='o')
plt.xlabel('Players')
plt.ylabel('Cumulative Runs')
plt.xticks(rotation=90)
plt.grid(True)
plt.show()
````

### Question 4 [Easy - 2D Line Plot]
Plot the number of TV Shows added to Netflix over the years using a green dashed line style with dot markers.

```python
# Write your code below
```

````python
plt.style.use('fivethirtyeight')
tvshow=netflix[netflix['type']=='TV Show'].groupby('release_year').size()

plt.figure(figsize=(25,7))
plt.plot(tvshow.index,tvshow.values,linestyle='dashed',marker='o',color='green',markersize=10)
plt.xticks(range(tvshow.index.min(), tvshow.index.max() + 1),rotation=90)
plt.title('Number of TV show per year')
plt.xlabel('Years')
plt.ylabel('Number of shows')
plt.grid(True)
plt.show()
````
